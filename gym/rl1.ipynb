```python
environment_name = 'CartPole-v1'
env = gym.make(environment_name)
```

# Describe environment

```python
env = gym.make("CartPole-v1", render_mode="rgb_array")

episodes = 5
for episode in range(1, episodes + 1):
    state, info = env.reset()
    done = False
    score = 0

    while not done:
        frame = env.render()
        action = env.action_space.sample()
        n_state, reward, terminated, truncated, info = env.step(action)
        done = terminated or truncated
        score += reward
        state = n_state
        print(f'Current State:{n_state}')
    print(f"Episode:{episode} Score:{score}")

env.close()
```

### Environment
- We start by defining our environment.
    - Add name `environment_name = "name"`
    - Make environment `env = gym.make(environment_name)`
-  Environment has an action space `env.action_space`
    - Which you may sample from it using `env.action_space.sample()`
- Environment has also the observation space `env.observation_space`
    - Which you can sample from `env.observation_space.sample()`


# Training

### Make your directories first

```python
log_patch = os.path.join('Training', 'Logs')
```

```python
env = gym.make(environment_name)
env = DummyVecEnv([lambda: env])
model = PPO('MlpPolicy', env, verbose=1, tensorboard_log=log_patch)
```

Using `DummyVecEnv` we wrap our environment in a dummy vector


```python
model.learn(total_timesteps=20000)
```

### Save model

```python
PPO_Path = os.path.join('Training', 'Saved_Models', 'PPO_Model_Cartpole')
model.save(PPO_Path)
```
